# Writing a class

A class describes a kind of object. Each object made from it is an
**instance**, with its own values.

```python
class Venue:
    def __init__(self, name, capacity):
        self.name = name
        self.capacity = capacity
        self.sold = 0


kiosque = Venue("Le Kiosque", 120)
casino = Venue("Grand Casino", 650)
print(kiosque.name, kiosque.capacity)
print(casino.name, casino.capacity)
```

`Venue("Le Kiosque", 120)` creates an empty object, then calls `__init__`
with it as `self`. `self.name = name` stores the value **in this object**.

## Methods

```python
class Venue:
    def __init__(self, name, capacity):
        self.name = name
        self.capacity = capacity
        self.sold = 0

    def seats_left(self):
        return self.capacity - self.sold

    def book(self, count):
        if count > self.seats_left():
            raise ValueError(f"Not enough seats: {self.seats_left()} left.")
        self.sold += count


kiosque = Venue("Le Kiosque", 120)
kiosque.book(4)
kiosque.book(10)
print(kiosque.sold, kiosque.seats_left())
```

`kiosque.book(4)` is a shorter way to write `Venue.book(kiosque, 4)`: the object before the dot becomes `self`.

```python
Venue.book(kiosque, 1)
print(kiosque.sold)
```

## Class attribute, instance attribute

```python
class Counter:
    created = 0

    def __init__(self, name):
        self.name = name
        self.value = 0
        Counter.created += 1

    def increment(self):
        self.value += 1


door_a = Counter("Door A")
door_b = Counter("Door B")
door_a.increment()
door_a.increment()
door_b.increment()
print(door_a.value, door_b.value)
print(Counter.created)
```

`value` belongs to each counter. `created` belongs to the class: one value, shared by all counters.

## Objects inside objects

```python
from datetime import datetime


class Show:
    def __init__(self, title, venue, start):
        self.title = title
        self.venue = venue
        self.start = start


casino = Venue("Grand Casino", 650)
gala = Show("Opening Gala", casino, datetime(2027, 6, 10, 20, 30))
print(gala.venue.name)
print(gala.start.hour)

gala.venue.book(2)
print(casino.sold)
```

`gala.venue` and `casino` are the same object: a change through one is seen through the other.
