# Special methods

Python calls the methods with double underscores by itself: `print` calls
`__str__`, `==` calls `__eq__`, `<` and `sorted` call `__lt__`.

```python
class Artist:
    def __init__(self, name):
        self.name = name


a = Artist("Nora Lebrun")
b = Artist("Nora Lebrun")
print(a)
print(a == b)
print(a == a)
```

Without `__eq__`, `==` only tells whether both names point to the **same object**.

```python
sorted([Artist("Sacha Diallo"), Artist("Max Pellerin")])
```

```python
class Artist:
    def __init__(self, name):
        self.name = name

    def __str__(self):
        return self.name

    def __repr__(self):
        return f"Artist({self.name!r})"

    def __eq__(self, other):
        if not isinstance(other, Artist):
            return NotImplemented
        return self.name.lower() == other.name.lower()

    def __lt__(self, other):
        return self.name.lower() < other.name.lower()
```

```python
nora = Artist("Nora Lebrun")
print(nora)
print(repr(nora))
print(f"On stage: {nora}")
```

```python
artists = [Artist("Sacha Diallo"), Artist("max pellerin"), Artist("Nora Lebrun")]
print(artists)
print(sorted(artists))
print(min(artists))
```

A list shows its items with `__repr__` (for the developer). `print` of a single object uses `__str__` (for the user).

```python
print(Artist("NORA LEBRUN") == nora)
print(Artist("nora lebrun") in artists)
print(nora == "Nora Lebrun")
```

`in` uses `__eq__`. Returning `NotImplemented` for another type lets Python answer `False`.

## A value with arithmetic

```python
class Duration:
    def __init__(self, minutes):
        self.minutes = minutes

    def __str__(self):
        return f"{self.minutes // 60}h{self.minutes % 60:02d}"

    def __add__(self, other):
        return Duration(self.minutes + other.minutes)


total = Duration(75) + Duration(20)
print(total)
```

| Written | Calls |
|---|---|
| `print(x)`, `str(x)`, `f"{x}"` | `x.__str__()` |
| `repr(x)`, a list containing `x` | `x.__repr__()` |
| `x == y`, `x in a_list` | `x.__eq__(y)` |
| `x < y`, `sorted`, `min`, `max` | `x.__lt__(y)` |
| `x + y` | `x.__add__(y)` |
| `len(x)` | `x.__len__()` |
