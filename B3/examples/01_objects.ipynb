# Using objects

Every value is an object: it has a type, **attributes** (values attached to
it) and **methods** (functions attached to it), reached with a dot.

```python
text = "  nora lebrun "
print(type(text))
print(text.title())
print(text.strip().title())
print(text)
```

A `str` method gives back a new text; `text` itself does not change. Methods
can be chained: each one works on the result of the previous one.

```python
print("  Nora   LEBRUN ".lower().split())
print(" ".join("  Nora   LEBRUN ".split()).title())
```

```python
acts = ["Nora", "Max"]
result = acts.append("Sacha")
print(acts)
print(result)
```

A `list` method changes the list itself, and gives back `None`.

```python
acts.upper()
```

## Dates: `datetime`

```python
from datetime import date, datetime, timedelta

opening = date(2027, 6, 10)
print(opening)
print(opening.year, opening.month, opening.day)
print(opening.weekday())
print(opening.strftime("%A %d %B %Y"))
```

```python
start = datetime(2027, 6, 11, 20, 30)
print(start)
print(start.hour, start.minute)
print(start.date())
print(start.isoformat())
```

From a text: `fromisoformat` is called on the class itself.

```python
print(datetime.fromisoformat("2027-06-11 20:30"))
print(date.fromisoformat("2027-06-11"))
```

```python
datetime.fromisoformat("11/06/2027 20:30")
```

## Computing with dates

```python
end = start + timedelta(minutes=75)
print(end)
print(end - start)
print((end - start).total_seconds() / 60)
print(start < end)
```

```python
wait = date(2027, 6, 10) - date(2026, 10, 1)
print(wait)
print(wait.days)
```

## Festival: a show after midnight

```python
late = datetime(2027, 6, 12, 0, 30)
print(late.date())
print((late - timedelta(hours=6)).date())
```

Moving the clock back 6 hours gives the evening the show belongs to.

## Formatting

| Code | Gives |
|---|---|
| `%Y-%m-%d` | `2027-06-11` |
| `%H:%M` | `20:30` |
| `%a %d %b` | `Fri 11 Jun` |
| `%A %d %B` | `Friday 11 June` |
