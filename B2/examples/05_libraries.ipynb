# Libraries

## The standard library

Installed with Python: `math`, `random`, `statistics`, `pathlib`, `csv`,
`json`, `datetime`... Documentation: <https://docs.python.org/3/library/>

````python
import math

print(math.sqrt(2))
print(math.ceil(650 / 40))
````

````python
from statistics import mean, median

durations = [120, 75, 90, 60, 130]
print(mean(durations))
print(median(durations))
````

````python
import random

random.seed(2027)
artists = ["Nora", "Max", "Sacha", "Inès"]
print(random.choice(artists))
random.shuffle(artists)
print(artists)
````

`random.seed` gives the same draw at every run: useful to test a program.

````python
import statistics as st

print(st.mean([12, 15]))
````

## External libraries

Published on <https://pypi.org>, installed with uv:

```bash
uv add rich
```

`uv add` downloads the library into `.venv/` and writes it in
`pyproject.toml`: `uv sync` installs it on any other computer. `rich` is
already installed in this course folder.

````python
from rich import print as rich_print

rich_print("[bold magenta]Deauville[/] Festival du [green]Rire[/]")
rich_print({"venue": "Le Kiosque", "capacity": 120})
````

````python
from rich.console import Console
from rich.table import Table

table = Table(title="Friday 11 June")
table.add_column("Time")
table.add_column("Venue")
table.add_column("Title", style="magenta")
table.add_row("18:00", "Théâtre des Planches", "Family Business")
table.add_row("20:30", "Grand Casino", "Tall Tales")
table.add_row("23:30", "Le Kiosque", "Late Laughs")

Console().print(table)
````

Before adding a library to a project: read its documentation (rich:
<https://rich.readthedocs.io>), check the date of the last release, and
prefer the standard library when it is enough.
