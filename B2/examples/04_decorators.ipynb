# Functions are values

A function can be put in a variable or given to another function.

````python
def shout(text: str) -> str:
    return text.upper() + "!"


speak = shout
print(speak("hello"))
print(shout)
````

````python
def apply_to_all(function, words: list[str]) -> list[str]:
    result = []
    for word in words:
        result.append(function(word))
    return result


print(apply_to_all(shout, ["ha", "hi"]))
print(apply_to_all(str.title, ["nora lebrun", "max pellerin"]))
````

## Sorting with `key=`

````python
shows = [
    {"title": "Late Laughs", "start": "23:30", "duration": 60},
    {"title": "Opening Gala", "start": "20:30", "duration": 120},
    {"title": "Improv Lab", "start": "15:00", "duration": 90},
]


def duration_of(show: dict) -> int:
    return show["duration"]


for show in sorted(shows, key=duration_of):
    print(show["duration"], show["title"])
````

````python
print(max(shows, key=duration_of)["title"])
for show in sorted(shows, key=duration_of, reverse=True):
    print(show["title"])
````

`key=duration_of`, without parentheses: `sorted` receives the function and
calls it on each show.

## Decorators

A decorator receives a function and gives back a new function that does
something more around it. `*args` accepts any number of arguments and passes
them on.

````python
def announce(function):
    def wrapper(*args):
        print(f"Calling {function.__name__}{args}")
        result = function(*args)
        print(f"{function.__name__} returned {result}")
        return result

    return wrapper


def add(a, b):
    return a + b


add = announce(add)
total = add(2, 3)
````

`@announce` above `def` is a shorter way to write `f = announce(f)`:

````python
@announce
def end_minutes(start: int, duration: int) -> int:
    return start + duration


end = end_minutes(1230, 90)
````

## A decorator of the standard library

````python
import time
from functools import cache


@cache
def slow_square(number: int) -> int:
    time.sleep(1)
    return number * number


start = time.perf_counter()
print(slow_square(12))
print(slow_square(12))
print(f"{time.perf_counter() - start:.1f} s")
````

The second call gives the saved result without running the function again.
