# Functions

`def` gives a name to a piece of code; calling the name runs it.

````python
def greet():
    print("Hello!")


greet()
greet()
````

## Parameters and return value

````python
def add_fees(price, fees):
    return price + fees


total = add_fees(22.0, 1.5)
print(total)
print(add_fees(12.0, 1.5) * 2)
````

````python
def show_total(price, fees):
    print(price + fees)


result = show_total(22.0, 1.5)
print(result)
````

`print` shows a value on the screen. `return` gives it back to the code that
called the function. Without `return`, a function gives `None`.

## Default values, named arguments

````python
def repeat(word, times=2, separator=" "):
    return separator.join([word] * times)


print(repeat("ha"))
print(repeat("ha", 4))
print(repeat("ha", separator="-", times=3))
````

## Type annotations and docstring

````python
def average(values: list[float]) -> float:
    """Average of the values, 0.0 for an empty list."""
    if len(values) == 0:
        return 0.0
    return sum(values) / len(values)


print(average([12.0, 15.5, 9.0]))
print(average([]))
````

````python
help(average)
````

Python does not check the annotations when the program runs. They tell the
reader (and VS Code) what the function expects and gives back.

## Local variables

````python
def double(number):
    result = number * 2
    return result


print(double(21))
````

````python
print(result)
````

`result` only exists inside `double`, while it runs.

## Returning two values

````python
def split_clock(clock: str) -> tuple[int, int]:
    hours, minutes = clock.split(":")
    return int(hours), int(minutes)


hours, minutes = split_clock("20:30")
print(hours)
print(minutes)
````

## Functions calling functions

````python
def to_minutes(clock: str) -> int:
    hours, minutes = split_clock(clock)
    return hours * 60 + minutes


def duration_between(start: str, end: str) -> int:
    return to_minutes(end) - to_minutes(start)


print(to_minutes("20:30"))
print(duration_between("20:30", "22:15"))
````

## A list given to a function

````python
def add_title(titles: list[str], title: str) -> None:
    titles.append(title)


programme = ["Opening Gala"]
add_title(programme, "Late Laughs")
print(programme)
````

The function receives the same list, not a copy: the change is visible
outside.

## Festival: functions on a show

````python
def describe(show: dict) -> str:
    return f"{show['start']} | {show['venue']} | {show['title']}"


def is_free(show: dict) -> bool:
    return show["price"] == 0


show = {"title": "Open Mic Night", "venue": "Le Kiosque", "start": "22:30", "price": 0.0}
print(describe(show))
print(is_free(show))
````

## Running a file, importing a file

In a `.py` file, the code that asks questions goes into `main()`:

```python
def main() -> None:
    name = input("Name: ")
    print(f"Hello {name}")


if __name__ == "__main__":
    main()
```

`uv run python file.py` runs `main()`. When the tests import the file to call
its functions, `__name__` is not `"__main__"`: nothing is asked.
