# Errors

## Reading a traceback

Start from the **last line**: type of error and message. The lines above show
the way from your call down to the line that failed.

````python
def to_minutes(clock):
    hours, minutes = clock.split(":")
    return int(hours) * 60 + int(minutes)


def end_minutes(start, duration):
    return to_minutes(start) + duration


end_minutes("20h30", 90)
````

| Error | Typical cause |
|---|---|
| `ValueError` | right type, wrong value: `int("abc")` |
| `TypeError` | wrong type: `"12" + 3` |
| `KeyError` | missing key in a dict |
| `IndexError` | position outside a list |
| `NameError` | variable not defined (or typo) |
| `FileNotFoundError` | wrong path |
| `ZeroDivisionError` | division by 0 |

## try / except

````python
text = "abc"
try:
    number = int(text)
    print(f"Got {number}")
except ValueError:
    print(f"'{text}' is not a whole number")
print("The program goes on")
````

## Asking again until the answer is valid

In a script, `answer = input(...)` replaces the list.

````python
answers = ["twelve", "-3", "12"]
for answer in answers:
    try:
        age = int(answer)
    except ValueError:
        print(f"{answer}: not a number")
        continue
    if age < 0:
        print(f"{answer}: cannot be negative")
        continue
    print(f"Age: {age}")
    break
````

## The error message

````python
try:
    int("22.5")
except ValueError as error:
    print("Message:", error)
````

## Several kinds of errors

````python
prices = {"full": 22.0, "reduced": 12.0, "guest": 0.0}
for category in ["full", "child", "guest"]:
    try:
        tickets = 100 / prices[category]
        print(f"{category}: {tickets:.1f} tickets for 100 EUR")
    except KeyError:
        print(f"{category}: unknown category")
    except ZeroDivisionError:
        print(f"{category}: free")
````

## Raising an error

````python
def check_duration(minutes: int) -> None:
    if minutes < 1 or minutes > 360:
        raise ValueError(f"A duration must be 1 to 360 minutes, got {minutes}.")


check_duration(90)
print("90 is fine")
````

````python
check_duration(500)
````

The function says what is wrong. The code that calls it decides what to do:

````python
for minutes in [90, 0, 120]:
    try:
        check_duration(minutes)
        print(f"{minutes}: ok")
    except ValueError as error:
        print(f"Refused: {error}")
````

## Replacing an error with a clearer one

````python
import json

text = '{"title": "Gala",'
try:
    data = json.loads(text)
except json.JSONDecodeError as error:
    raise ValueError("The programme file is broken.") from error
````

`from error` keeps the original error in the traceback, above the new one.

## What not to do

```python
try:
    ...
except:      # catches everything, even a typo in a variable name
    pass     # and says nothing
```

Catch the error you expect (`ValueError`, `KeyError`...) and do something
useful with it: ask again, skip the line with a message, stop with a clear
explanation.
