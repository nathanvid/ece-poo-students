# Conditions

The indented block runs only when the condition is `True`.

```python
number = 7

if number % 2 == 0:
    print(f"{number} is even.")
else:
    print(f"{number} is odd.")
```

## `if / elif / else`

The first `True` branch wins, the others are skipped.

```python
number = -4

if number > 0:
    print("positive")
elif number < 0:
    print("negative")
else:
    print("zero")
```

## Comparisons

| `==` | `!=` | `<` | `<=` | `>` | `>=` |
|---|---|---|---|---|---|
| equal | different | less | less or equal | greater | greater or equal |

`=` stores a value, `==` compares two values.

```python
print(5 == 5.0)
print("Nora" == "nora")
print("apple" < "banana")
print(3 != 4)
```

```python
a = 12
b = 30

if a > b:
    largest = a
else:
    largest = b
print(f"Largest: {largest}")
```

```python
age = 20
if age >= 18
    print("adult")
```

## `and`, `or`, `not`

```python
print(True and False)
print(True or False)
print(not True)
```

A range check reads like maths:

```python
age = 20
print(18 <= age <= 25)
```

## Festival: who may attend a 16+ show

```python
age = 14
has_ticket = True
min_age = 16

if not has_ticket:
    print("Please go to the box office first.")
elif age >= min_age:
    print("Enjoy the show!")
else:
    print(f"Sorry, this show is for {min_age}+ only.")
```

## Festival: ticket category

The order of the branches matters.

```python
age = 70

if age < 12:
    category = "child"
elif age < 26:
    category = "young"
elif age >= 65:
    category = "senior"
else:
    category = "adult"
print(category)
```
