# Variables and types

A variable is a **name** for a value.

```python
x = 7
y = 2
print(x + y)
```

```python
x = x + 1
print(x)
x += 1
print(x)
```

Two variables at once, and swapping them:

```python
a, b = 1, 2
a, b = b, a
print(a, b)
```

## Numbers: `int` and `float`

`/` always gives a float. `//` is the whole division, `%` the remainder.

```python
print(7 + 2, 7 - 2, 7 * 2)
print(7 / 2)
print(7 // 2)
print(7 % 2)
print(7**2)
```

```python
print(round(3.14159, 2))
print(abs(-5))
```

## Text: `str`

```python
first_name = "Nora"
last_name = "Lebrun"
full_name = first_name + " " + last_name
print(full_name)
print(len(full_name))
```

```python
print("ha" * 3)
```

## Booleans: `bool`

```python
is_open = True
print(is_open)
print(3 > 2)
print(3 == 2)
```

## `type()`

```python
print(type(x))
print(type(7 / 2))
print(type(full_name))
print(type(is_open))
```

## Names

`snake_case`, and a name that says what the value is: `duration_minutes` rather than `d`.

## f-strings

```python
duration_minutes = 75
hours = duration_minutes // 60
minutes = duration_minutes % 60
print(f"The show lasts {hours}h{minutes:02d}")
```

`:02d` at least 2 digits, `:.2f` 2 decimals, `:<12` and `:>12` align on 12 characters.

```python
ticket_price = 22.5
print(f"Price: {ticket_price:.2f} EUR")
print(f"|{'Title':<12}|{'Venue':>12}|")
```

## Conversions

`"12"` is text, `12` is a number.

```python
print("12" + "3")
print(12 + 3)
```

```python
print("12" + 3)
```

```python
seats = int("180")
print(seats + 20)
price = float("22.5")
print(price * 2)
print("Edition " + str(2027))
```

`int()` cuts the decimals, it does not round:

```python
print(int(7.9))
```

```python
int("22.5")
```

`bool()`: `0`, `0.0` and `""` are `False`, everything else is `True`.

```python
print(bool(0), bool(3), bool(""), bool("no"))
```

## Festival: a box office total

```python
full_price = 22.0
tickets = 3
total = tickets * full_price
print(f"{tickets} tickets: {total:.2f} EUR")
```
