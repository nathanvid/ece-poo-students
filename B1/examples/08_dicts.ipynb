# Dictionaries

A dict links **keys** to values.

```python
translations = {"hello": "bonjour", "cat": "chat", "laugh": "rire"}
print(translations["laugh"])
print(len(translations))
```

```python
print("dog" in translations)
print(translations.get("dog"))
print(translations.get("dog", "?"))
```

```python
print(translations["dog"])
```

## Adding, changing, removing

```python
translations["dog"] = "chien"
translations["hello"] = "salut"
del translations["cat"]
print(translations)
```

## Looping

```python
for english in translations:
    print(english)
```

```python
for english, french in translations.items():
    print(f"{english} -> {french}")
```

## Counting

```python
letters = {}
for letter in "hahaha hihi":
    if letter == " ":
        continue
    if letter in letters:
        letters[letter] += 1
    else:
        letters[letter] = 1
print(letters)
```

Same thing, shorter, with `.get()`:

```python
counts = {}
for letter in "hahaha":
    counts[letter] = counts.get(letter, 0) + 1
print(counts)
```

## Festival: one show

```python
show = {
    "title": "Nothing Serious",
    "artist": "Nora Lebrun",
    "venue": "Grand Casino",
    "day": "2027-06-11",
    "start": "20:30",
    "duration": 75,
}
show["price"] = 22.0

for key, value in show.items():
    print(f"{key:>10}: {value}")
```

## Festival: a programme, list of dicts

```python
programme = [
    {"title": "Nothing Serious", "venue": "Grand Casino", "duration": 75},
    {"title": "Open Mic Night", "venue": "Le Kiosque", "duration": 120},
    {"title": "Improv Lab", "venue": "Le Kiosque", "duration": 90},
]

for item in programme:
    if item["venue"] == "Le Kiosque":
        print(item["title"])
```

Count the shows per venue:

```python
shows_per_venue = {}
for item in programme:
    venue = item["venue"]
    shows_per_venue[venue] = shows_per_venue.get(venue, 0) + 1
print(shows_per_venue)
```

Find the longest show: keep the best one seen so far.

```python
longest = programme[0]
for item in programme:
    if item["duration"] > longest["duration"]:
        longest = item
print(f"{longest['title']} ({longest['duration']} min)")
```
