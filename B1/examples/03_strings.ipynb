# Text

Positions start at **0**.

```python
word = "Deauville"
print(len(word))
print(word[0])
print(word[-1])
```

```python
print(word[0:3])
print(word[3:])
```

```python
print(word[20])
```

## Methods

A method is an action the text knows how to do: `text.method()`.

```python
print(word.upper())
print(word.lower())
print("  hello  ".strip())
print("I love Paris".replace("Paris", "Deauville"))
```

```python
print("comedy".startswith("com"))
print("banana".count("a"))
```

A method returns a **new** text. The original does not change:

```python
name = "nora"
name.upper()
print(name)
name = name.upper()
print(name)
```

Methods can be chained, from left to right:

```python
answer = "  YES "
print(answer.strip().lower())
```

## Searching and checking

```python
print("ville" in word)
print("x" in word)
print(word.find("v"))
```

```python
print("2027".isdigit())
print("20h30".isdigit())
print("   ".strip() == "")
```

## Splitting and joining

```python
print("the show must go on".split())
print("20:30".split(":"))
print("-".join(["2027", "06", "10"]))
```

## Looping over characters

```python
for letter in "Rire":
    print(letter)
```
