# Loops

## `for` and `range`

`range(5)` gives 0 to 4: the end is excluded.

```python
for i in range(5):
    print(i)
```

```python
for i in range(1, 6):
    print(i, end=" ")
```

```python
for i in range(0, 20, 5):
    print(i, end=" ")
```

```python
for i in range(5, 0, -1):
    print(i, end=" ")
```

```python
for i in range(3):
print(i)
```

## Accumulating

A variable that grows at each turn.

```python
total = 0
for i in range(1, 101):
    total += i
print(total)
```

```python
n = 7
for i in range(1, 11):
    print(f"{n} x {i} = {n * i}")
```

## Looping over a text

```python
vowels = 0
for letter in "deauville festival":
    if letter in "aeiouy":
        vowels += 1
print(vowels)
```

## `while`

Repeats as long as the condition is `True`. Something inside the loop must change the condition.

```python
power = 1
while power < 1000:
    power *= 2
print(power)
```

## `break` and `continue`

`continue` skips to the next turn, `break` leaves the loop.

```python
for i in range(1, 10):
    if i == 3:
        continue
    if i == 6:
        break
    print(i, end=" ")
```

## Festival: countdown

```python
for minutes_left in range(3, 0, -1):
    print(f"Show starts in {minutes_left} min...")
print("Curtain up!")
```

## Festival: total stage time

```python
act_durations = [12, 15, 10, 20]
stage_time = 0
for duration in act_durations:
    stage_time += duration
print(f"Total stage time: {stage_time} min")
```
