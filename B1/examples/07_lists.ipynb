# Lists

A list stores several values **in order**.

```python
numbers = [4, 8, 15, 16, 23, 42]
print(numbers)
print(len(numbers))
```

## Reading

Positions start at **0**. A negative position counts from the end.

```python
print(numbers[0])
print(numbers[-1])
print(numbers[1:3])
```

```python
print(numbers[10])
```

## Changing

```python
fruits = ["apple", "banana", "cherry"]
fruits.append("kiwi")
fruits.insert(0, "mango")
fruits.remove("banana")
print(fruits)
```

Replace by position, or remove the last one with `pop()`:

```python
fruits[1] = "apricot"
last = fruits.pop()
print(last)
print(fruits)
```

## Asking questions

```python
print("cherry" in fruits)
print(fruits.index("cherry"))
print(sum(numbers), min(numbers), max(numbers))
print(sorted([3, 1, 2]))
```

## Looping

```python
for fruit in fruits:
    print(fruit.upper())
```

`enumerate` gives the position **and** the value:

```python
for position, fruit in enumerate(fruits, start=1):
    print(f"{position}. {fruit}")
```

## Building a list in a loop

Start from an empty list, `append` inside the loop.

```python
squares = []
for i in range(1, 6):
    squares.append(i * i)
print(squares)
```

```python
evens = []
for n in numbers:
    if n % 2 == 0:
        evens.append(n)
print(evens)
```

## From text to list, and back

```python
words = "to be or not to be".split()
print(words)
print(" ".join(words[:2]))
```

## A list of lists

`grid[row][column]`

```python
grid = [[1, 2, 3], [4, 5, 6]]
print(grid[1][0])
```

## Festival: a running order

```python
lineup = ["Nora Lebrun", "Max Pellerin", "Sacha Diallo"]
lineup.append("Inès Moreau")
lineup.insert(0, "Paul Kerbrat")
lineup.remove("Max Pellerin")

for number, act in enumerate(lineup, start=1):
    print(f"{number}. {act}")
```

```python
position = lineup.index("Sacha Diallo") + 1
print(f"Sacha Diallo is act number {position}")
```
