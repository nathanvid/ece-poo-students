# Metaclasses (opening, not graded)

Classes are objects too. Their type is `type`.

```python
class Show:
    pass


print(type(3))
print(type("Nora"))
print(type(Show))
print(type(type))
```

```python
def shout(self):
    return "HELLO!"


Dynamic = type("Dynamic", (), {"greeting": "hello", "shout": shout})
print(Dynamic().greeting, Dynamic().shout())
```

`type(name, parents, attributes)` builds a class, exactly like a `class` statement. A **metaclass** is a subclass of `type`: it can act each time a class is created.

```python
class Registered(type):
    kinds = {}

    def __new__(mcs, name, bases, namespace):
        cls = super().__new__(mcs, name, bases, namespace)
        if "kind" in namespace:
            Registered.kinds[namespace["kind"]] = cls
        return cls


class Performance(metaclass=Registered):
    pass


class SoloShow(Performance):
    kind = "solo"


class Workshop(Performance):
    kind = "workshop"


print(Registered.kinds)
print(Registered.kinds["solo"])
```

The same registry, without a metaclass, with `__init_subclass__` (called each time a subclass is created):

```python
class Performance:
    kinds = {}

    def __init_subclass__(cls):
        super().__init_subclass__()
        Performance.kinds[cls.kind] = cls


class SoloShow(Performance):
    kind = "solo"


class Lineup(Performance):
    kind = "lineup"


print(Performance.kinds)
```

Pydantic, SQLModel or Django read the fields of a class when it is created,
with these tools. In the project, an `if kind == "solo"` in
`serialization.py` stays clearer.
