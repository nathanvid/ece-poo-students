# Inheritance

A subclass gets everything from its parent class, then adds or replaces.

```python
class Shape:
    def __init__(self, name):
        self.name = name

    def area(self):
        return 0.0

    def describe(self):
        return f"{self.name}: {self.area():.2f}"


class Rectangle(Shape):
    def __init__(self, width, height):
        super().__init__("rectangle")
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height


class Square(Rectangle):
    def __init__(self, side):
        super().__init__(side, side)
        self.name = "square"


for shape in [Shape("point"), Rectangle(4, 3), Square(2)]:
    print(shape.describe())
```

`describe` is written once, in `Shape`, and calls the `area` of the real object: this is **polymorphism**.

```python
square = Square(2)
print(isinstance(square, Square), isinstance(square, Rectangle), isinstance(square, Shape))
print(issubclass(Square, Shape))
print(type(square).__name__)
```

## Forgetting `super().__init__`

```python
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14159 * self.radius**2


print(Circle(1).describe())
```

`Shape.__init__` never ran: the circle has no `name`.

## Extending a method of the parent

```python
class Stage(Rectangle):
    def describe(self):
        return super().describe() + " m² of stage"


print(Stage(8, 5).describe())
```

## Festival: three kinds of performances

```python
from datetime import datetime, timedelta


class Performance:
    kind = "performance"

    def __init__(self, title, venue, start, duration):
        self.title = title
        self.venue = venue
        self.start = start
        self.duration = duration

    @property
    def end(self):
        return self.start + timedelta(minutes=self.duration)

    def artists(self):
        raise NotImplementedError

    def __str__(self):
        return f"{self.start:%H:%M}-{self.end:%H:%M} {self.title} ({', '.join(self.artists())})"


class SoloShow(Performance):
    kind = "solo"

    def __init__(self, title, venue, start, duration, artist):
        super().__init__(title, venue, start, duration)
        self.artist = artist

    def artists(self):
        return [self.artist]


class Lineup(Performance):
    kind = "lineup"

    def __init__(self, title, venue, start, duration, host, acts):
        super().__init__(title, venue, start, duration)
        self.host = host
        self.acts = acts

    def artists(self):
        return [self.host] + self.acts


programme = [
    SoloShow("Late Laughs", "Le Kiosque", datetime(2027, 6, 11, 23, 30), 60, "Nora Lebrun"),
    Lineup("Opening Gala", "Grand Casino", datetime(2027, 6, 10, 20, 30), 120,
           "Paul Kerbrat", ["Nora Lebrun", "Max Pellerin"]),
]
for performance in programme:
    print(performance.kind, "|", performance)
```

`end` and `__str__` are written once; `artists()` depends on the kind.

Inheritance fits when "a `SoloShow` **is a** `Performance`" is true. A festival
**has** performances: that is an attribute, not inheritance.

```python
Performance("?", "Le Kiosque", datetime(2027, 6, 11, 20, 0), 60).artists()
```
